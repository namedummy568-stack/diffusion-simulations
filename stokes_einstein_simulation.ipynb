```python
import numpy as np
import matplotlib.pyplot as plt

def stokes_einstein_diffusion_coefficient(k_B, T, eta, r):
    """
    Calculates the diffusion coefficient using the Stokes-Einstein equation.

    Parameters:
    k_B (float): Boltzmann constant (J/K)
    T (float): Absolute temperature (K)
    eta (float): Dynamic viscosity of the liquid (Pa·s)
    r (float): Hydrodynamic radius of the spherical particle (m)

    Returns:
    float: Diffusion coefficient (m^2/s)
    """
    D = (k_B * T) / (6 * np.pi * eta * r)
    return D

# Constants
k_B = 1.380649e-23  # Boltzmann constant (J/K)

# Simulation parameters
temperature = 298.15  # K (25 °C)
viscosity_water = 8.9e-4  # Pa·s (viscosity of water at 25 °C)
particle_radii = np.linspace(1e-9, 10e-9, 100)  # 1 nm to 10 nm

# Calculate diffusion coefficients
diffusion_coefficients = [
    stokes_einstein_diffusion_coefficient(k_B, temperature, viscosity_water, r)
    for r in particle_radii
]

# Plotting the results
plt.figure(figsize=(10, 6))
plt.plot(particle_radii * 1e9, np.array(diffusion_coefficients) * 1e10)
plt.xlabel("Particle Radius (nm)")
plt.ylabel("Diffusion Coefficient (x10^-10 m^2/s)")
plt.title("Diffusion Coefficient vs. Particle Radius (Stokes-Einstein)")
plt.grid(True)
plt.show()
```
